### 1.3 Explanation of Your Model

> 參考連結 https://blogs.novita.ai/zh-TW/introducing-roberta-base-model-a-comprehensive-overview/
![](./pics/introducing-roberta-base-model-a-comprehensive-overview-0_cSe_fTJLx2FSVzuS.png)

RoBERTa 的架構與 BERT 類似，但為了提高效能，作者對架構和訓練程式進行了一些簡單的設計變更。這些變化包括：
1. 刪除下一句預測（NSP）目標：在 BERT 中，模型使用輔助 NSP 損失進行訓練，以預測文件的兩個片段是來自相同還是不同的文件。作者對有和沒有 NSP 損失的模型版本進行了實驗，發現消除​​ NSP 損失可以匹配或略微提高下游任務的性能。
2. 使用更大的批量大小和更長的序列進行訓練：BERT 最初訓練了 1 萬步，批量大小為 256 個序列。 RoBERTa 經過 125 步（每步 2,000 個序列）和 31,000 步（每批 8,000 個序列）的訓練。更大的批次可以提高掩蔽語言建模目標的困惑度和最終任務的準確性。使用分散式並行訓練也更容易實現並行化。
3. 動態改變掩蔽模式：在 BERT 中，掩蔽在資料預處理期間完成一次，從而產生單一靜態掩蔽。為了避免這種情況，訓練資料被複製並在 10 個時期內使用不同的策略屏蔽 40 次，最終有 4 個時期使用相同的屏蔽。該策略與動態屏蔽進行了比較，每次將資料傳遞到模型時都會產生不同的屏蔽。

## 1. Preprocessing Steps

````python
import os
import polars as pl

os.chdir('/home/poyu39/github/poyu39/DM2025-Kaggle')

pl.Config.set_tbl_cols(None)
pl.Config.set_tbl_rows(10)
pl.Config.set_fmt_str_lengths(100)

# MODEL_NAME = 'j-hartmann/emotion-english-distilroberta-base'
# OUTPUT_DIR = 'outputs/emotion-english-distilroberta-base'
MODEL_NAME = 'j-hartmann/emotion-english-roberta-large'
OUTPUT_DIR = 'outputs/emotion-english-roberta-large'

MAX_LEN = 512
BATCH_SIZE = 16
EPOCHS = 4
SEED = 39
LR = 1e-5
````

````python
import torch
import numpy as np

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
````

````python
import json
import re

post = []

for item in json.load(open('datasets/final_posts.json', 'r')):
    post.append(item.get('root').get('_source').get('post'))

def clean_text(text):
    
    # 移除 URL
    text = str(text)
    text = re.sub(r'http\S+|www\S+', '', text)
    
    # 移除 emoji
    # emoji_pattern = re.compile(
    #     '['
    #     '\U0001F600-\U0001F64F'
    #     '\U0001F300-\U0001F5FF'
    #     '\U0001F680-\U0001F6FF'
    #     '\U0001F1E0-\U0001F1FF'
    #     ']+', flags=re.UNICODE
    # )
    # text = emoji_pattern.sub('', text)
    
    # 去除多餘空白
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

post_df = pl.DataFrame(post)
post_df = post_df.with_columns(pl.col('text').cast(pl.Utf8).alias('text'))
post_df = post_df.with_columns(
    pl.col('text').map_elements(clean_text, return_dtype=pl.Utf8).alias('text')
)
post_df
````

````python
emotion_df = pl.read_csv('datasets/emotion.csv')
emotion_df = emotion_df.rename({'id': 'post_id'})
emotion_df
````

````python
split_df = pl.read_csv('datasets/data_identification.csv')
split_df = split_df.rename({'id': 'post_id'})
split_df
````

````python
train_idx = split_df.filter(pl.col('split') == 'train')['post_id'].to_list()
test_idx = split_df.filter(pl.col('split') == 'test')['post_id'].to_list()

train_df = post_df.filter(pl.col('post_id').is_in(train_idx)).join(emotion_df, on='post_id')
test_df = post_df.filter(pl.col('post_id').is_in(test_idx))
test_df = test_df.with_columns(pl.lit('').alias('emotion'))

val_df = train_df.sample(fraction=0.2, seed=SEED)
train_df = train_df.filter(~pl.col('post_id').is_in(val_df['post_id'].to_list()))

display(train_df)
display(val_df)
display(test_df)
````

````python
labels = train_df.select(pl.col("emotion")).unique().to_series().to_list()
labels = sorted(labels)
label2id = {l: i for i, l in enumerate(labels)}
id2label = {i: l for l, i in label2id.items()}
display(label2id)
display(id2label)
````

## 2. Feature Engineering Steps

````python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments, EarlyStoppingCallback

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)
trainer = Trainer(model=model)
````

````python
class EmotionsDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels=None):
        self.encodings = encodings
        self.labels = labels
    def __len__(self):
        return len(next(iter(self.encodings.values())))
    def __getitem__(self, idx):
        item = {k: torch.tensor(v[idx]) for k, v in self.encodings.items()}
        if self.labels is not None:
            item['labels'] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item
````

````python
def tokenize_series(series):
        return tokenizer(series.to_list(), truncation=True, padding='max_length', max_length=MAX_LEN)

train_enc = tokenize_series(train_df['text'])
val_enc = tokenize_series(val_df['text'])

train_labels = [label2id[e] for e in train_df['emotion'].to_list()]
val_labels = [label2id[e] for e in val_df['emotion'].to_list()]

train_dataset = EmotionsDataset(train_enc, train_labels)
val_dataset = EmotionsDataset(val_enc, val_labels)
````

## 3. Model Implementation Steps

````python
from sklearn.metrics import accuracy_score, f1_score

def compute_metrics(pred):
    labels = pred.label_ids
    preds = np.argmax(pred.predictions, axis=1)
    return {
        'accuracy': accuracy_score(labels, preds),
        'f1_micro': f1_score(labels, preds, average='micro'),
        'f1_macro': f1_score(labels, preds, average='macro'),
    }
````

````python
training_args = TrainingArguments(
        output_dir=OUTPUT_DIR,
        num_train_epochs=EPOCHS,
        learning_rate=LR,
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE * 2,
        eval_strategy='steps',
        eval_steps=200,
        save_strategy='steps',
        save_steps=200,
        logging_strategy='steps',
        logging_steps=200,
        metric_for_best_model='f1_macro',
        greater_is_better=True,
        seed=SEED,
        fp16=torch.cuda.is_available(),
        push_to_hub=False,
    )
trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        processing_class=tokenizer,
        compute_metrics=compute_metrics,
        callbacks=[EarlyStoppingCallback(early_stopping_patience=2)],
    )
trainer.train()
trainer.save_model(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print('Saved model to', OUTPUT_DIR)
````

````python
# inference

import torch.nn.functional as F

model = AutoModelForSequenceClassification.from_pretrained(f'{OUTPUT_DIR}/checkpoint-1800')
tokenizer = AutoTokenizer.from_pretrained(f'{OUTPUT_DIR}/checkpoint-1800')
trainer = Trainer(model=model, processing_class=tokenizer)

test_enc = tokenize_series(test_df['text'])
test_dataset = EmotionsDataset(test_enc)

raw_preds = trainer.predict(test_dataset)

probs = F.softmax(torch.tensor(raw_preds.predictions), dim=1).numpy()
pred_ids = probs.argmax(axis=1)
pred_labels = [id2label[i] for i in pred_ids]


result_df = test_df.with_columns([
    pl.Series('emotion', pred_labels),
    pl.Series('prob', probs.max(axis=1))
])

result_df = result_df.rename({'post_id': 'id'})

result_df.select(['id', 'emotion']).write_csv(f'{OUTPUT_DIR}/submission.csv')
print('Inference done', f'{OUTPUT_DIR}/submission.csv')
````
